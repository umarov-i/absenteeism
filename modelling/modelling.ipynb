### Scaling

```python
# from sklearn.preprocessing import StandardScaler

# scaler = StandardScaler()
```

```python
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.preprocessing import StandardScaler

class CustomScaler(BaseEstimator, TransformerMixin):
    def __init__(self, columns, copy=True, with_mean=True, with_std=True):
        self.scaler = StandardScaler(
            copy=copy,
            with_mean=with_mean,
            with_std=with_std
        )
        self.columns = columns
        self.copy = copy
        self.with_mean = with_mean
        self.with_std = with_std
        self.mean_ = None
        self.var_ = None

    def fit(self, X, y=None):
        self.scaler.fit(X[self.columns], y)
        self.mean_ = np.mean(X[self.columns])
        self.var_ = np.var(X[self.columns])
        return self

    def transform(self, X, y=None, copy=None):
        init_col_order = X.columns
        X_scaled = pd.DataFrame(self.scaler.transform(X[self.columns]), columns=self.columns)
        X_not_scaled = X.loc[:, ~X.columns.isin(self.columns)]
        return pd.concat([X_not_scaled, X_scaled], axis=1)[init_col_order]
```

```python
unscaled_inputs.columns.values
```

```python
# columns_to_scale = ['Month Value','Day of the Week','Transportation Expense','Distance to Work',
# 'Age', 'Daily Work Load Average','Body Mass Index','Children','Pet']
columns_to_omit = ['Reason 1', 'Reason 2', 'Reason 3', 'Reason 4', 'Education']
```

```python
columns_to_scale = [x for x in unscaled_inputs.columns.values if x not in columns_to_omit]
```

```python
scaler = CustomScaler(columns_to_scale)
```

```python
scaler.fit(unscaled_inputs)
```

```python
scaled_inputs = scaler.transform(unscaled_inputs)
scaled_inputs
```

```python
scaled_inputs.shape
```

### Splitting

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(scaled_inputs, targets, test_size=0.2, random_state=42)
```

```python
print(x_train.shape, y_train.shape)
```

### Modelling

```python
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
```

```python
reg = LogisticRegression()
reg.fit(x_train, y_train)
```

```python
reg.score(x_train, y_train)
```

#### Checking

```python
model_outputs = reg.predict(x_train)
model_outputs
```

```python
y_train
```

```python
print(np.sum(model_outputs == y_train), '\n', np.sum(model_outputs == y_train) / model_outputs.shape[0])
```

```python
print(reg.intercept_, reg.coef_)
```

```python
feature_names = unscaled_inputs.columns.values
```

```python
summary_table = pd.DataFrame(columns=['Feature Name'], data=feature_names)
summary_table['coefs'] = np.transpose(reg.coef_)

summary_table
```

```python
summary_table.index = summary_table.index + 1
summary_table.loc[0] = ['intercept', reg.intercept_[0]]
summary_table = summary_table.sort_index()
summary_table
```

```python
summary_table['odds ratio'] = np.exp(summary_table.coefs)
summary_table = summary_table.sort_values('odds ratio', ascending=False)
summary_table
```
